```python
import pandas as pd
from pathlib import Path
import os
import numpy as np
import csv
import argparse

folder = 'iran2'

tweetid = []
userid = []
account_date = []
account_lang = []
tweet_lang = []
tweet_text = []
tweet_time = []
tweet_client = []
reply_user_id = []
quote_tweet_id = []
is_retweet = []
retweet_user_id = []
hashtags = []
mentions = []



for file in Path(folder).iterdir():
    if file.name.endswith('.csv'):
        print('Reading {}'.format(file))
        with open(file, 'r', encoding="utf8") as f:
            csv_f = csv.reader((line.replace('\0','') for line in f), quotechar='"') # added to combat _csv.Error: line contains NULL byteprint('womp womp')
            for row in csv_f:
                if(len(row) < 1):
                    continue
                print(row[0])
                tweetid.append(row[0])
                userid.append(row[1])
                account_date.append(row[9])
                account_lang.append(row[10])
                tweet_lang.append(row[11])
                tweet_text.append(row[12])
                tweet_time.append(row[13])
                tweet_client.append(row[14])
                reply_user_id.append(row[15])
                quote_tweet_id.append(row[17])
                is_retweet.append(row[18])
                retweet_user_id.append(row[19])
                hashtags.append(row[27])
                mentions.append(row[29])
```
