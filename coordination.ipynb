```python
## Create combo list of all outside contacts for given country and compare
## against master_uids list for coordination

folder = 'russia' # country being examined


outsideRT = pd.read_csv('{}_outsideCountryRT_ids_256188'.format(folder))
outsideRT['type'] = 'RT'
outsideRT = outsideRT.rename(columns={'retweetid': 'id'})
outsideRT = outsideRT.applymap(str)
outsideRply = pd.read_csv('{}_outsideCountryRply_ids_597499'.format(folder))
outsideRply['type'] = 'Rply'
outsideRply = outsideRply.rename(columns={'replyid': 'id'})
outsideRply = outsideRply.applymap(str)
outsideIDs = [outsideRT, outsideRply]
outsideDF = pd.concat(outsideIDs)

#perform comparison btw master and country specific lists
coordination = pd.merge(outsideDF, master_uids, on=['id'], how='inner')
coordination = coordination.drop(coordination.columns[[0]], axis=1)

print(coordination.head())
print('There are {} outside connections for {}'.format(len(coordination), folder))

coordination.to_csv('{}_outsideConnections_{}.csv'.format(folder,len(coordination)), index=False)
```
