```python
country_list = ['ecuador', 'china', 'cuba', 'egypt', 'honduras', 'indonesia', 'iran',
       'russia', 'saudi_arabia', 'serbia', 'turkey', 'uae', 'uae_egypt', 'venezuela']
```

```python
for country in country_list: 
    records = pd.read_csv('coordinationExternal_{}.csv'.format(country))  # open file
    countryUIDs = records[['userid']].drop_duplicates() # extract unique userID list
    print('{}: {}'.format(country, len(countryUIDs)))
    allRecords = pd.read_pickle('{}_parsed.pkl'.format(country))  # read old completed parsed file
    uidRecords = allRecords.merge(countryUIDs, left_on='userid', right_on='userid') # join with UID list for all records for country
    uidRecords.to_csv('finalRecords_{}.csv'.format(country), index=False)
    print(country)
    print(min(uidRecords.tweet_time))
    print(max(uidRecords.tweet_time))
```

```python
country = 'cuba'

records = pd.read_csv('finalRecords_{}.csv'.format(country))
records['userid'] = records['userid'].apply(str)


records['TweetTime'] = [x[:10] for x in records['tweet_time']] # keep first 10 chars of date for pivot (yyyy-mm-dd)
 
import sys

import datetime

start = datetime.datetime.strptime("2008-09-15", "%Y-%m-%d")
end = datetime.datetime.strptime("2020-05-15", "%Y-%m-%d")
date_generated = [start + datetime.timedelta(days=x) for x in range(0, (end-start).days)]

date_list = []

for date in date_generated:
    date_list.append(date.strftime("%Y-%m-%d"));
#     print date.strftime("%m/%d/%Y")
# print(date_list)



df = pd.DataFrame(date_list, columns =['TweetTime'])
df['userid'] = 'ignoreMe'

records = records.append(df)

df = pd.DataFrame(records, columns=["userid", "TweetTime"])
table = pd.pivot_table(df, values='userid', index=['userid'],
                       columns=['TweetTime'], aggfunc=len, fill_value=0)
table.to_csv('frequency_{}.csv'.format(country))

len(table)
```
